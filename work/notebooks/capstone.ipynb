## 1. Question

*The research question and the decision it supports.*

The research question is:

- Which existing content pages should the product team prioritize for refresh to maximize the chance of reversing observed search decline while preserving remaining demand?

This notebook supports the product decision by:

- describing the page-level evidence of decline,
- explaining how the ranking is meant to identify the best refresh candidates,
- and making the recommendation clearer for a refresh planning process.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

The data source is the anonymized starter sample provided in `data/raw/content_refresh_anonymized.csv`.
This sample represents page-level search and engagement statistics for existing content, with no direct personal or client identifiers.

Key points about the data:

- The release is a bundled anonymized starter dataset, not a live production feed.
- The data covers existing content pages that have measurable search traffic.
- Pages with zero recent impressions are not useful for refresh ranking and are excluded from the feature-preparation set.
- Pages with very recent content age are also excluded because they have not had enough time to show a stable decline pattern.
- Duplicate page identifiers are removed so the analysis is based on unique content units.

This section is public-safe because it describes the dataset and preprocessing in general terms only.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

Assumptions:

- Search decline can be detected from page-level trend signals in the available anonymized data.
- Pages with ongoing impressions are the most valuable candidates for refresh because they still have search demand.
- A practical refresh ranking should balance decline risk and remaining visibility.

Features and label definition:

- The model uses features such as recent impressions, change in position, engagement metrics, and content age.
- The label is a binary indication of whether a page is showing downward trend signals in its search performance.
- The target is defined to capture pages that are declining now but still have enough demand to warrant a refresh.

Baseline and validation:

- The baseline is a simple rule-based ranking that uses existing signals to approximate refresh priority.
- The comparison is done on the same validation split so the model performance is judged against the baseline fairly.
- Validation is designed to avoid using future information in the training period, which is critical for honest evaluation.

Leakage checks:

- The methodology includes explicit checks to avoid using data that would only be available after the refresh decision is made.
- The evaluation is based on held-out slices of the sample, not the same rows used to train the model.
- This helps ensure the results are indicative of real decision-support value.

## 5. Limitations

*What this work cannot claim.*

This work is a decision-support analysis, not a guarantee of outcome.

What it cannot claim:

- It cannot prove that a refresh will definitively reverse search decline for an individual page.
- It cannot capture every external factor, such as search engine algorithm changes or broader market shifts.
- It does not replace editorial judgment; it only highlights candidates with the strongest signals.
- It is based on anonymized sample data, so it may not reflect a full production environment.

The appropriate use is to guide review priority, not to automate refresh execution.
